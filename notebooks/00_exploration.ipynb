```python
def parse_credit_history(x):
    if pd.isna(x):
        return np.nan

    try:
        parts = x.split()
        years = int(parts[0])
        months = int(parts[3])
        return years + months / 12
    except:
        return np.nan
```

```python
train_df["Credit_History_Age"] = train_df["Credit_History_Age"].apply(
    parse_credit_history
)
train_df["Credit_History_Age"].describe()
```
