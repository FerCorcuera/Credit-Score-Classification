## Tasks:

- Convert the months into numbers to split the data temporally
- Ignore identifiers for now that does not provide important informationl ike name ssn,


```python
month_mapping = {
    'January' : 202201,
    'February' : 202202,
    'March' : 202203,
    'April' : 202204,
    'May' : 202205,
    'June' : 202206,
    'July' : 202207,
    'August' : 202208,

}

train_df['num_month'] = train_df['Month'].map(month_mapping)

train_df['num_month'].value_counts()
```
